```python
import os 
import pandas as pd
```

```python
def piRNA_lendist(file) : 

    seqlens = []
    firsts = []
    rpms = []
    features = []
    genes = []
    
    with open(file, 'r') as f : 
        for line in f : 
            if not line.startswith("chrom") :
                info = line.strip().split("\t")
                if info[7] == 'piRNA' : 
                    seq = str(info[4])
                    seql = len(seq)
                    first = seq[0]
                    rpm = float(info[15])
                    gene = info[10]

                    seqlens.append(seql)
                    firsts.append(first)
                    rpms.append(rpm)
                    features.append(info[7])
                    genes.append(gene)
    f.close()

    df = pd.DataFrame({
        'seqlen':seqlens,
        'firstnt':firsts,
        'rpm':rpms,
        'feature' : features,
        'gene' : genes
    })  
    
    return df
```
